```python
# Step 38K-1: Check exact training-weather timestamp matches

# Make sure weather is loaded
weather = pd.read_csv(
    "../data/raw/power-laws-forecasting-energy-consumption-weather.csv",
    sep=";"
)

weather["Timestamp"] = pd.to_datetime(
    weather["Timestamp"],
    errors="coerce",
    utc=True
)

# Make sure integrated training timestamp is datetime
train_integrated["Timestamp"] = pd.to_datetime(
    train_integrated["Timestamp"],
    errors="coerce",
    utc=True
)

# Unique keys only - avoids unnecessary large merge
train_keys = train_integrated[
    ["SiteId", "Timestamp"]
].drop_duplicates()

weather_keys = weather[
    ["SiteId", "Timestamp"]
].drop_duplicates()

matched_weather = train_keys.merge(
    weather_keys,
    on=["SiteId", "Timestamp"],
    how="inner"
)

print("Unique training SiteId + Timestamp keys:",
      len(train_keys))

print("Unique weather SiteId + Timestamp keys:",
      len(weather_keys))

print("Exact matching keys:",
      len(matched_weather))

print(
    "Exact weather match percentage:",
    round(
        len(matched_weather) /
        len(train_keys) * 100,
        2
    ),
    "%"
)
```

```python
# Step 38K-4: Nearest weather timestamp analysis

# -----------------------------
# 1. Prepare training timestamps
# -----------------------------

train_time_check = (
    train_integrated[
        ["SiteId", "Timestamp"]
    ]
    .drop_duplicates()
    .rename(columns={"Timestamp": "TrainTimestamp"})
)

# -----------------------------
# 2. Prepare weather timestamps
# -----------------------------

weather_time_check = (
    weather[
        ["SiteId", "Timestamp"]
    ]
    .drop_duplicates()
    .rename(columns={"Timestamp": "WeatherTimestamp"})
)

# -----------------------------
# 3. Sort primarily by time
# -----------------------------

train_time_check = train_time_check.sort_values(
    ["TrainTimestamp", "SiteId"]
).reset_index(drop=True)

weather_time_check = weather_time_check.sort_values(
    ["WeatherTimestamp", "SiteId"]
).reset_index(drop=True)

# -----------------------------
# 4. Find nearest weather record
# -----------------------------

nearest_weather = pd.merge_asof(
    train_time_check,
    weather_time_check,
    left_on="TrainTimestamp",
    right_on="WeatherTimestamp",
    by="SiteId",
    direction="nearest",
    tolerance=pd.Timedelta("2 hours")
)

# -----------------------------
# 5. Calculate time difference
# -----------------------------

nearest_weather["TimeGap"] = (
    nearest_weather["TrainTimestamp"]
    - nearest_weather["WeatherTimestamp"]
).abs()

# -----------------------------
# 6. Basic results
# -----------------------------

print(
    "Total training SiteId + Timestamp keys:",
    len(nearest_weather)
)

matched_count = nearest_weather["TimeGap"].notna().sum()

print(
    "\nTraining keys with weather within 2 hours:",
    matched_count
)

print(
    "\nPercentage with weather within 2 hours:",
    round(
        matched_count / len(nearest_weather) * 100,
        2
    ),
    "%"
)

# -----------------------------
# 7. Time-gap statistics
# -----------------------------

print("\nTime-gap statistics:")

print(
    nearest_weather["TimeGap"]
    .dropna()
    .describe()
)

# -----------------------------
# 8. Time-gap categories
# -----------------------------

gap_data = nearest_weather["TimeGap"].dropna()

gap_categories = pd.cut(
    gap_data,
    bins=[
        pd.Timedelta("0 min"),
        pd.Timedelta("5 min"),
        pd.Timedelta("10 min"),
        pd.Timedelta("15 min"),
        pd.Timedelta("30 min"),
        pd.Timedelta("1 hour"),
        pd.Timedelta("2 hours")
    ],
    include_lowest=True
)

print("\nTime-gap categories:")

print(
    gap_categories
    .value_counts()
    .sort_index()
)
```

```python
# Step 40-5: Create rolling consumption features

# Make sure data is correctly ordered
model_data = model_data.sort_values(
    ["SiteId", "Timestamp"]
).reset_index(drop=True)

# Create a separate source containing only past observations
rolling_source = model_data[
    ["SiteId", "Timestamp", "Value"]
].copy()

# Set timestamp as index for time-based rolling
rolling_source = rolling_source.set_index("Timestamp")

# 24-hour rolling mean
rolling_24h_mean = (
    rolling_source
    .groupby("SiteId")["Value"]
    .rolling("24h", closed="left")
    .mean()
    .reset_index()
)

rolling_24h_mean = rolling_24h_mean.rename(
    columns={"Value": "Rolling_24h_Mean"}
)

# 24-hour rolling standard deviation
rolling_24h_std = (
    rolling_source
    .groupby("SiteId")["Value"]
    .rolling("24h", closed="left")
    .std()
    .reset_index()
)

rolling_24h_std = rolling_24h_std.rename(
    columns={"Value": "Rolling_24h_Std"}
)

# 7-day rolling mean
rolling_7d_mean = (
    rolling_source
    .groupby("SiteId")["Value"]
    .rolling("7d", closed="left")
    .mean()
    .reset_index()
)

rolling_7d_mean = rolling_7d_mean.rename(
    columns={"Value": "Rolling_7d_Mean"}
)

# Merge rolling features
model_data = model_data.merge(
    rolling_24h_mean,
    on=["SiteId", "Timestamp"],
    how="left"
)

model_data = model_data.merge(
    rolling_24h_std,
    on=["SiteId", "Timestamp"],
    how="left"
)

model_data = model_data.merge(
    rolling_7d_mean,
    on=["SiteId", "Timestamp"],
    how="left"
)

# Check results
print("Model data shape:", model_data.shape)

print("\nRolling 24h mean available:",
      model_data["Rolling_24h_Mean"].notna().sum())

print("Rolling 24h std available:",
      model_data["Rolling_24h_Std"].notna().sum())

print("Rolling 7d mean available:",
      model_data["Rolling_7d_Mean"].notna().sum())

print("\nAvailability percentages:")

for col in [
    "Rolling_24h_Mean",
    "Rolling_24h_Std",
    "Rolling_7d_Mean"
]:
    print(
        col,
        ":",
        round(
            model_data[col].notna().mean() * 100,
            2
        ),
        "%"
    )

print("\nSample:")
print(
    model_data[
        [
            "SiteId",
            "Timestamp",
            "Value",
            "Value_1h_Ago",
            "Value_24h_Ago",
            "Value_7d_Ago",
            "Rolling_24h_Mean",
            "Rolling_24h_Std",
            "Rolling_7d_Mean"
        ]
    ].head(10)
)
```

```python
# Step 42: Train and save the baseline forecasting model

from sklearn.ensemble import RandomForestRegressor
import numpy as np
import pandas as pd
import joblib
import os

feature_columns = [
    "SiteId",
    "Surface",
    "Sampling",
    "BaseTemperature",
    "MondayIsDayOff",
    "TuesdayIsDayOff",
    "WednesdayIsDayOff",
    "ThursdayIsDayOff",
    "FridayIsDayOff",
    "SaturdayIsDayOff",
    "SundayIsDayOff",
    "IsHoliday",
    "Hour",
    "DayOfWeek",
    "Day",
    "Month",
    "Year",
    "IsWeekend",
    "Value_1h_Ago",
    "Value_24h_Ago",
    "Value_7d_Ago",
    "Rolling_24h_Mean",
    "Rolling_24h_Std",
    "Rolling_7d_Mean"
]

target_column = "Value"

# Prepare training and validation data
X_train = train_data[feature_columns].copy()
X_validation = validation_data[feature_columns].copy()

y_train = train_data[target_column].copy()
y_validation = validation_data[target_column].copy()

# Convert boolean columns to integers
boolean_columns = [
    "MondayIsDayOff",
    "TuesdayIsDayOff",
    "WednesdayIsDayOff",
    "ThursdayIsDayOff",
    "FridayIsDayOff",
    "SaturdayIsDayOff",
    "SundayIsDayOff"
]

for col in boolean_columns:
    X_train[col] = X_train[col].astype("int8")
    X_validation[col] = X_validation[col].astype("int8")

# Fill missing feature values
X_train = X_train.fillna(0)
X_validation = X_validation.fillna(0)

# Log-transform target
y_train_log = np.log1p(y_train)

# Create model
model = RandomForestRegressor(
    n_estimators=100,
    max_depth=20,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

print("Training model...")
print("Training rows:", len(X_train))
print("Validation rows:", len(X_validation))

# Train
model.fit(
    X_train,
    y_train_log
)

print("Model training completed.")

# Save trained model
model_path = "../models/random_forest_energy_model.pkl"

joblib.dump(model, model_path)

print("Model saved to:", model_path)

# Generate validation predictions
print("Generating validation predictions...")

predictions_log = model.predict(X_validation)

predictions = np.expm1(predictions_log)
predictions = np.maximum(predictions, 0)

print("Prediction completed.")

# Save validation predictions
prediction_output = validation_data[
    ["SiteId", "Timestamp", "ForecastId", "Value", "IsHoliday"]
].copy()

prediction_output = prediction_output.rename(
    columns={"Value": "Actual"}
)

prediction_output["Predicted"] = predictions

prediction_path = "../outputs/validation_predictions.csv"

prediction_output.to_csv(
    prediction_path,
    index=False
)

print("Predictions saved to:", prediction_path)

print("\nFirst 10 predictions:")
print(predictions[:10])

print("\nFirst 10 actual values:")
print(y_validation.iloc[:10].values)

print("\nFiles created successfully:")
print(model_path)
print(prediction_path)
```

```python
# Step 43: Evaluate the forecasting model

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Actual and predicted values
actual = y_validation.to_numpy()
predicted = predictions

# -----------------------------------------
# MAE
# -----------------------------------------
mae = mean_absolute_error(actual, predicted)

# -----------------------------------------
# RMSE
# -----------------------------------------
rmse = np.sqrt(
    mean_squared_error(actual, predicted)
)

# -----------------------------------------
# R²
# -----------------------------------------
r2 = r2_score(actual, predicted)

# -----------------------------------------
# sMAPE
# -----------------------------------------
denominator = (
    np.abs(actual) +
    np.abs(predicted)
)

smape_mask = denominator != 0

smape = np.mean(
    2 * np.abs(
        predicted[smape_mask] -
        actual[smape_mask]
    ) /
    denominator[smape_mask]
) * 100

# -----------------------------------------
# WAPE
# -----------------------------------------
wape = (
    np.sum(np.abs(actual - predicted)) /
    np.sum(np.abs(actual))
) * 100

# -----------------------------------------
# Print results
# -----------------------------------------

print("========== FORECASTING RESULTS ==========")

print(f"MAE   : {mae:,.2f}")
print(f"RMSE  : {rmse:,.2f}")
print(f"sMAPE : {smape:.2f}%")
print(f"WAPE  : {wape:.2f}%")
print(f"R²    : {r2:.4f}")

print("\n==========================================")

# -----------------------------------------
# Sample comparison
# -----------------------------------------

comparison = pd.DataFrame({
    "Actual": actual[:20],
    "Predicted": predicted[:20],
    "Absolute_Error": np.abs(
        actual[:20] - predicted[:20]
    )
})

print("\nFirst 20 actual vs predicted:")
print(comparison)
```

```python
# Step 44-1: Analyze forecasting errors for anomaly detection

import numpy as np
import pandas as pd

# Create result dataframe
results = validation_data[
    [
        "SiteId",
        "Timestamp",
        "ForecastId",
        "Value",
        "IsHoliday"
    ]
].copy()

results = results.rename(
    columns={"Value": "Actual"}
)

results["Predicted"] = predictions

# -----------------------------------------
# Absolute error
# -----------------------------------------

results["Absolute_Error"] = (
    np.abs(
        results["Actual"] -
        results["Predicted"]
    )
)

# -----------------------------------------
# Percentage deviation
# -----------------------------------------

denominator = results["Actual"].abs()

results["Deviation_Percent"] = np.where(
    denominator > 0,
    results["Absolute_Error"] /
    denominator * 100,
    0
)

# -----------------------------------------
# Error distribution
# -----------------------------------------

print("========== ERROR DISTRIBUTION ==========")

print(
    results["Deviation_Percent"].describe(
        percentiles=[
            0.50,
            0.75,
            0.90,
            0.95,
            0.99
        ]
    )
)

# -----------------------------------------
# Large deviations
# -----------------------------------------

print("\n========== DEVIATION THRESHOLDS ==========")

for threshold in [10, 20, 30, 50, 75, 100]:
    count = (
        results["Deviation_Percent"] >= threshold
    ).sum()

    percentage = (
        count / len(results) * 100
    )

    print(
        f">= {threshold}% : "
        f"{count:,} rows "
        f"({percentage:.2f}%)"
    )

# -----------------------------------------
# Top anomalous observations
# -----------------------------------------

print("\n========== TOP 10 DEVIATIONS ==========")

print(
    results[
        [
            "SiteId",
            "Timestamp",
            "Actual",
            "Predicted",
            "Deviation_Percent",
            "IsHoliday"
        ]
    ]
    .sort_values(
        "Deviation_Percent",
        ascending=False
    )
    .head(10)
)
```

```python
# Step 44-2: Create anomaly and deviation categories

# -----------------------------------------
# Direction of deviation
# -----------------------------------------

results["Deviation"] = (
    results["Actual"] -
    results["Predicted"]
)

results["Deviation_Direction"] = np.where(
    results["Deviation"] > 0,
    "Higher than predicted",
    np.where(
        results["Deviation"] < 0,
        "Lower than predicted",
        "As predicted"
    )
)

# -----------------------------------------
# Avoid unreliable percentage deviations
# for extremely small actual values
# -----------------------------------------

MIN_CONSUMPTION = 100

results["Reliable_Deviation"] = (
    results["Actual"] >= MIN_CONSUMPTION
)

# -----------------------------------------
# Anomaly flag
# -----------------------------------------

results["Anomaly"] = (
    (results["Reliable_Deviation"]) &
    (results["Deviation_Percent"] >= 50)
)

# -----------------------------------------
# Deviation category
# -----------------------------------------

results["Deviation_Category"] = np.select(
    [
        results["Deviation_Percent"] < 20,
        results["Deviation_Percent"] < 50,
        results["Deviation_Percent"] >= 50
    ],
    [
        "Normal",
        "Moderate deviation",
        "Potential anomaly"
    ],
    default="Normal"
)

# -----------------------------------------
# Summary
# -----------------------------------------

print("========== ANOMALY SUMMARY ==========")

print(
    "Total validation observations:",
    len(results)
)

print(
    "Potential anomalies:",
    results["Anomaly"].sum()
)

print(
    "Potential anomaly percentage:",
    round(
        results["Anomaly"].mean() * 100,
        2
    ),
    "%"
)

print("\nDeviation categories:")

print(
    results["Deviation_Category"]
    .value_counts()
)

print("\nDeviation direction:")

print(
    results["Deviation_Direction"]
    .value_counts()
)

# -----------------------------------------
# Top potential anomalies
# -----------------------------------------

print("\n========== TOP POTENTIAL ANOMALIES ==========")

anomalies = (
    results[
        results["Anomaly"]
    ]
    .sort_values(
        "Deviation_Percent",
        ascending=False
    )
)

print(
    anomalies[
        [
            "SiteId",
            "Timestamp",
            "Actual",
            "Predicted",
            "Deviation",
            "Deviation_Percent",
            "Deviation_Direction",
            "IsHoliday"
        ]
    ].head(20)
)
```

```python
# Step 45: Prepare structured results for IBM Granite

# Select the final anomaly records
granite_data = results[
    results["Anomaly"] == True
].copy()

# Keep the most useful fields for explanation
granite_data = granite_data[
    [
        "SiteId",
        "Timestamp",
        "Actual",
        "Predicted",
        "Deviation",
        "Deviation_Percent",
        "Deviation_Direction",
        "Deviation_Category",
        "IsHoliday"
    ]
].copy()

# -----------------------------------------
# Add simple contextual interpretation
# -----------------------------------------

granite_data["Context"] = np.where(
    granite_data["IsHoliday"] == 1,
    "Holiday period",
    "Non-holiday period"
)

# -----------------------------------------
# Create JSON-style records
# -----------------------------------------

granite_records = granite_data.head(100).to_dict(
    orient="records"
)

print("Total anomaly records available:",
      len(granite_data))

print("\nRecords prepared for Granite:",
      len(granite_records))

print("\nExample Granite input:")

print(granite_records[0])
```
